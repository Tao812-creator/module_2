##measures of correlation
correlation
covariance= measures whether two variables move together: positive move in the same direction, while -ve move in the opposite direction

pearson correlation coefficient(r)- standardizes the covariance to always fall btwn -1 and +1
-1 = perfect negative correlation
0 - no linear correlation
+1= perfect positive correlation

CORELATION DOES NOT MEAN CAUSATION- causation implies that one variable directly affects the other

```python
import numpy as np


rng = np.random.default_rng(42) #seed 

n = 25

study_hours = rng.uniform(1, 10, n).round(1)

#exam_score
exam_score = (40 + study_hours + 5 + rng.normal(0, 8, n)).round(1)
exam_score = np.clip(exam_score, 0, 100)

age = rng.integers(18, 45, n)


class_df = pd.DataFrame({
    'study hours':study_hours,
    'exam_score':exam_score,
    'age':age}
)

class_df.head()
```

```python
#COVARIANCE BTWN STUDY HOURS AND EXAM SCORES

cov_matrix = class_df[['study hours', 'exam_score']].cov()

cov_matrix
```

```python
corr_study_score = class_df['study hours'].corr(class_df['exam_score'])

corr_study_score
```

```python
corr_age_score = class_df['age'].corr(class_df['exam_score'])

corr_age_score
```

```python
#scatter plot
import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
plt.scatter(class_df['study hours'],class_df['exam_score'])
plt.title('study hours  vs exam score')
plt.xlabel('study_hours')
plt.ylabel('exam score')
plt.show()
```

```python
#histogram
plt.figure(figsize=(6,4))
plt.hist(class_df['exam_score'], bins=8)
plt.title('Distribution of exam score')
plt.xlabel('exam score')
plt.ylabel('no. of students')
plt.show()

```

```python
import seaborn as sns
plt.figure(figsize=(4,3))

sns.heatmap(class_df.corr(),annot=True, cmap='coolwarm',vmin=-1,vmax=1)
plt.title('correlation heatmap')
plt.show
```
